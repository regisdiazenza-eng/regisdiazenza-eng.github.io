## Ping vers le VLAN 800 avant et après la configuration du NAT et de l'ip route

On remarque que la connexion au reseau extérieur (172.31.25.9) a échoué car l'adresse IP source correspond à une adresse privée  qui n'est pas reconnu sur internet.En l'absence du nat sur le routeur, cette adresse n'est pas traduite en adresse publique, les paquets sont donc rejetés à l'extérieur du réseau ce qui explique l'absence de réponses ICMP sur la capture.

Pour avoir accès à Internet nous avons fait des ACL, configuré le NAT et configuré une route par défaut
