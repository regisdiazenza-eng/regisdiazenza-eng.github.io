# Seance 4 - SAE 12 : Mise en œuvre d'une maquette réseau virtualisée

## Objectif
L’objectif de cette séance est de mettre en œuvre la maquette réseau en utilisant uniquement des machines virtuelles et en deployant les services suivants sur des machines virtuelles :
- Serveur Web (Apache)
- Serveur DHCP
- Serveur DNS
- Vérifier leur bon fonctionnement par des tests réseau et captures de trames

---

## Déploiement du Serveur Web (Apache)

### 3.1 Installation du service
Le serveur Apache a été installé sur la VM serveur Ubuntu pour héberger le service Web :
``
sudo apt update
sudo apt install apache2 -y
`` 
#### Cette commande permet de verifier si apache est active
``
sudo systemctl status apache2
``

#### Demarrer Apache
``
sudo systemctl start apache2
``
#### Test d’accès Web

Depuis un navigateur :

http://192.168.10.20

Le serveur Apache affiche la page par défaut.
Les captures d’écran d’installation et de test seront ajoutées dans une version finale du compte rendu.

## Installation du serveur DHCP

### Configuration d'une IP statique 

```
sudo nano /etc/netplan/01-netcfg.yaml

network:
  version: 2
  ethernets:
    enp0s3:
      dhcp4: no
      addresses:
        - 192.168.10.10/24
```
