# Expressions régulières 

Signification des symboles :

- "." Un caractère qui n'est pas une New Line
- ^ Début de la chaîne de caractère
- $ Fin de la chaîne
- "*" 0 ou plusireus répétitions du motif à gauche 
- "+" 1 ou plusieurs répétitions du motif à gauche
- "?" 0 ou 1 répétition du motif à gauche
- { } nombre de répétitions du motif à gauche
- [ ] Matche un des motifs entre crochets
- \ crée des raccourcis (\d pour les chiffres, \w pour les mots)
- | Ou
- ( ) Groupement

```python
print (re.match(r"GR(.)?S", "GRIS"))
```

```python
import re 
def check_num(num):
    
    l = num.split(" ")
    if len(l) == 5:
        if re.match(r"(0\d )((\d{2} ){3})(\d{2})", num) :
            value = "TRUE"
        else :
            value = "False"
    else :
        value = "FALSE"
    return value

print(check_num("07 46 28 93 55"))
```
