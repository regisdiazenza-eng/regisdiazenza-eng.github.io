program_pi.py

```python
from math import sqrt

def mypi(x):
    list_s = list()
    for i in range (1, x+1):
        list_s.append(1/(i ** 2))
    pi = sqrt(sum(list_s) * 6)
    return pi
if __name__ == "main":
    n = int(input("n = "))
    pi = mypi(n)
    print(pi)
```

program_pi_principal.py

```python
from program_pi import mypi
from math import pi


n = int(input("n = "))
if n != 0:
    nb = mypi(n)
    erreur = ((abs(nb - pi) / pi ) * 100 )
    print("Le nombre trouvé est :",nb)
    print("La valeur de pi est :", pi)
    print("L'erreur relative est :", erreur, "%")
else:
    print("Veuillez entrer un nombre entier différent de 0")
```

scopes.py

```python
x = "valeur globale"
# définition d'une fonction qui renvoit la valeur de x
def f1() :
    # indication de l'appel à la fonction f1
	print ("--Appel f1()")
	# donne la valeur de x prenant en compte la fonction f1
	print (" x dans f1 :", x)
 
 
#définition d'une fonction qui renvoit une valeur de x définie dans celle-ci
def f2() :
    # indication de l'appel à la fonction f2
	print ("--Appel f2()")
	# valeur attribuée à x 
	x = "valeur locale"
	# donne la valeur de x prenant en compte la fonction f2
	print (" x dans f2 :", x)
 
 
# définition d'une fonction qui renvoit les valeurs de x et y
def f3() :
    # indication de l'appel à la fonction f3
	print ("--Appel f3()")
	global x
	# valeur attribuée à x 
	x = "valeur modifie"
	# valeur attribuée à y
	y = 12
	# affiche la valeur de x dans la fonction f3
	print (" x dans f3 :", x)
	# affiche la valeur de y dans la fonction f3
	print (" y dans f3 :", y)

# affiche la valeur initiale de x 
print ("x au niveau global :", x)
# appel de la fonction
f1()
# print de value of x
print ("x au niveau global :", x)
f2()
print ("x au niveau global :", x)
f3()
print ("x au niveau global :", x)
print ("y au niveau global :", y)
# The code crashes at the last line because y is only defined in the function f3
```
